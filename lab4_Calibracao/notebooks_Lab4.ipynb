Nesta etapa, foi executado o programa L4_cal.py dentro da pasta do laboratório para realizar a calibração utilizando as imagens de exemplo fornecidas no material.


```python
import cv2
import numpy as np
import os
import glob
 
# Defining the dimensions of checkerboard
CHECKERBOARD = (6,8)
criteria = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 30, 0.001)
 
# Creating vector to store vectors of 3D points for each checkerboard image
objpoints = []
# Creating vector to store vectors of 2D points for each checkerboard image
imgpoints = [] 
 
 
# Defining the world coordinates for 3D points
objp = np.zeros((1, CHECKERBOARD[0] * CHECKERBOARD[1], 3), np.float32)
objp[0,:,:2] = np.mgrid[0:CHECKERBOARD[0], 0:CHECKERBOARD[1]].T.reshape(-1, 2)
prev_img_shape = None
 
# Extracting path of individual image stored in a given directory
images = glob.glob('*.jpg')
for fname in images:
    img = cv2.imread(fname)
    gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
    # Find the chess board corners
    # If desired number of corners are found in the image then ret = true
    ret, corners = cv2.findChessboardCorners(gray, CHECKERBOARD, cv2.CALIB_CB_ADAPTIVE_THRESH + cv2.CALIB_CB_FAST_CHECK + cv2.CALIB_CB_NORMALIZE_IMAGE)
     
    """
    If desired number of corner are detected,
    we refine the pixel coordinates and display 
    them on the images of checker board
    """
    if ret == True:
        objpoints.append(objp)
        # refining pixel coordinates for given 2d points.
        corners2 = cv2.cornerSubPix(gray, corners, (11,11),(-1,-1), criteria)
         
        imgpoints.append(corners2)
 
        # Draw and display the corners
        img = cv2.drawChessboardCorners(img, CHECKERBOARD, corners2, ret)
     
    #cv2.imshow('img',img)
    #cv2.waitKey(0)
 
cv2.destroyAllWindows()
 
h,w = img.shape[:2]
 
"""
Performing camera calibration by 
passing the value of known 3D points (objpoints)
and corresponding pixel coordinates of the 
detected corners (imgpoints)
"""
ret, mtx, dist, rvecs, tvecs = cv2.calibrateCamera(objpoints, imgpoints, gray.shape[::-1], None, None)
 
print("Camera matrix : \n")
print(mtx)
print("dist : \n")
print(dist)
print("rvecs : \n")
print(rvecs)
print("tvecs : \n")
print(tvecs)
```

## Interpretação dos parâmetros
    • K: é a matriz de parâmetros intrínsecos da câmera. Ela reúne informações como distância focal em pixels nas direções horizontal e vertical, posição do ponto principal e, quando aplicável, o fator de skew. Essa matriz descreve como a câmera projeta os pontos 3D no plano da imagem.
    • R: é a matriz de rotação que descreve a orientação da câmera em relação ao sistema de coordenadas do padrão observado. Ela informa como os eixos da câmera estão orientados em relação ao objeto de calibração.
    • t: é o vetor de translação, que indica a posição relativa entre a câmera e o padrão de calibração. Em conjunto com R, ele compõe os parâmetros extrínsecos.
    • dist: é o vetor dos coeficientes de distorção da lente, usado para modelar deformações ópticas, principalmente a distorção radial e, em alguns casos, a distorção tangencial.

O fato de resultar um conjunto de R e t para cada imagem ocorre porque, em cada captura, o tabuleiro de calibração aparece em uma posição e orientação diferentes em relação à câmera. Assim, os parâmetros intrínsecos são únicos para a câmera utilizada, mas os parâmetros extrínsecos mudam de uma imagem para outra, pois descrevem a pose relativa entre a câmera e o padrão em cada aquisição.

Nesse contexto, R representa a rotação do sistema de coordenadas do padrão em relação ao sistema de coordenadas da câmera, enquanto t representa a translação entre esses sistemas. Em outras palavras, essas variáveis informam como o tabuleiro estava posicionado e orientado no espaço no momento em que cada imagem foi capturada. Por isso, ao utilizar várias imagens do mesmo padrão em poses diferentes, obtém-se um conjunto de matrizes R e vetores t, um para cada observação usada no processo de calibração.

## Descrição do procedimento

Após a obtenção dos parâmetros de calibração da câmera, foram selecionadas imagens coloridas capturadas com os dispositivos calibrados para a etapa de correção de distorção. Inicialmente, esses parâmetros foram carregados no notebook Jupyter, incluindo a matriz intrínseca da câmera e os coeficientes de distorção. Em seguida, foi utilizada a função cv.getOptimalNewCameraMatrix() para calcular uma nova matriz ajustada, com base em um parâmetro de escala livre, a fim de otimizar a área útil da imagem corrigida.

Na sequência, a correção de distorção foi realizada por dois métodos distintos. No primeiro, utilizou-se a função cv.undistort(), que aplica diretamente os parâmetros de calibração para gerar a imagem corrigida. No segundo, utilizou-se a abordagem de remapeamento com cv.remap(), que exige o cálculo prévio dos mapas de transformação e depois aplica esses mapas à imagem original. Os dois resultados foram armazenados e organizados para comparação visual lado a lado com as imagens originais.

Como solicitado no roteiro, o desenvolvimento dessa etapa foi documentado em notebook Jupyter, com registro do código-fonte, das imagens utilizadas e das saídas produzidas. Para a análise, devem ser incluídas pelo menos duas imagens por aluno, sendo uma obtida com a webcam e outra com uma câmera diferente, permitindo observar de forma prática os efeitos da correção da distorção radial em diferentes dispositivos.

```python
import cv2
import numpy as np
import os
import glob
 
# Defining the dimensions of checkerboard
CHECKERBOARD = (6,8)
criteria = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 30, 0.001)
 
# Creating vector to store vectors of 3D points for each checkerboard image
objpoints = []
# Creating vector to store vectors of 2D points for each checkerboard image
imgpoints = [] 
 
 
# Defining the world coordinates for 3D points
objp = np.zeros((1, CHECKERBOARD[0] * CHECKERBOARD[1], 3), np.float32)
objp[0,:,:2] = np.mgrid[0:CHECKERBOARD[0], 0:CHECKERBOARD[1]].T.reshape(-1, 2)
prev_img_shape = None
 
# Extracting path of individual image stored in a given directory
images = glob.glob('*.jpeg')
for fname in images:
    img = cv2.imread(fname)
    gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
    # Find the chess board corners
    # If desired number of corners are found in the image then ret = true
    ret, corners = cv2.findChessboardCorners(gray, CHECKERBOARD, cv2.CALIB_CB_ADAPTIVE_THRESH + cv2.CALIB_CB_FAST_CHECK + cv2.CALIB_CB_NORMALIZE_IMAGE)
     
    """
    If desired number of corner are detected,
    we refine the pixel coordinates and display 
    them on the images of checker board
    """
    if ret == True:
        objpoints.append(objp)
        # refining pixel coordinates for given 2d points.
        corners2 = cv2.cornerSubPix(gray, corners, (11,11),(-1,-1), criteria)
         
        imgpoints.append(corners2)
 
       
 
ret, mtx, dist, rvecs, tvecs = cv2.calibrateCamera(objpoints, imgpoints, gray.shape[::-1], None, None)

cv2.destroyAllWindows()

img = cv2.imread('selfie.jpeg')
h,  w = img.shape[:2]
newcameramtx, roi = cv2.getOptimalNewCameraMatrix(mtx, dist, (w,h), 1, (w,h))
# undistort
dst = cv2.undistort(img, mtx, dist, None, newcameramtx)
 
# crop the image
x, y, w, h = roi
dst = dst[y:y+h, x:x+w]
cv2.imwrite('selfie_calibresult_undisort.png', dst)

# undistort
mapx, mapy = cv2.initUndistortRectifyMap(mtx, dist, None, newcameramtx, (w,h), 5)
dst = cv2.remap(img, mapx, mapy, cv2.INTER_LINEAR)
 
# crop the image
x, y, w, h = roi
dst = dst[y:y+h, x:x+w]
cv2.imwrite('selfie_calibresult_remapping.png', dst)

h,w = img.shape[:2]
 
"""
Performing camera calibration by 
passing the value of known 3D points (objpoints)
and corresponding pixel coordinates of the 
detected corners (imgpoints)
"""
ret, mtx, dist, rvecs, tvecs = cv2.calibrateCamera(objpoints, imgpoints, gray.shape[::-1], None, None)
 
print("Camera matrix : \n")
print(mtx)
print("dist : \n")
print(dist)
print("rvecs : \n")
print(rvecs)
print("tvecs : \n")
print(tvecs)
```

## Análise dos resultados
Na comparação entre as imagens originais e as imagens corrigidas, observou-se a redução do efeito de curvatura das linhas retas nas regiões periféricas da imagem, evidenciando a compensação da distorção radial introduzida pela lente. Após a correção, elementos geométricos como bordas, superfícies retangulares e alinhamentos lineares passaram a apresentar representação visual mais compatível com a geometria real da cena, indicando que os parâmetros estimados na calibração foram adequados para o processo de retificação.

Em relação aos métodos empregados, cv.undistort() e cv.remap() produziram resultados visualmente semelhantes quanto à remoção da distorção, uma vez que ambos utilizam os mesmos parâmetros intrínsecos e coeficientes estimados na calibração. Entretanto, o método baseado em remapeamento mostrou-se mais flexível para aplicações em que os mapas de correção podem ser previamente calculados e reutilizados, o que pode ser vantajoso em rotinas repetitivas ou em processamento de múltiplas imagens. De modo geral, a comparação confirmou que a calibração da câmera não apenas permite estimar parâmetros geométricos do sistema de aquisição, mas também viabiliza melhorias práticas na qualidade métrica e visual das imagens obtidas.

## Análise e Discussão
Os estudos realizados ao longo deste laboratório permitiram relacionar, de forma prática, os conceitos teóricos da geometria de câmeras com resultados experimentais obtidos por meio do OpenCV. No experimento A, utilizando imagens de exemplo, foi possível observar uma situação de calibração mais controlada, em que os parâmetros obtidos serviram como referência para compreender a estrutura das saídas do algoritmo e o significado das matrizes e vetores associados ao processo.

No experimento B, a calibração da webcam evidenciou de maneira mais clara a influência das condições reais de aquisição sobre os resultados. Fatores como enquadramento, qualidade do foco, uniformidade da iluminação, resolução da câmera e variedade de poses do tabuleiro impactam diretamente a estabilidade da detecção dos pontos e, consequentemente, a precisão dos parâmetros estimados. Essa etapa mostrou que a calibração não depende apenas do algoritmo empregado, mas também da qualidade do conjunto de imagens utilizado.

Já no experimento C, a utilização de uma segunda câmera pessoal permitiu comparar dispositivos distintos e verificar que cada câmera possui comportamento óptico próprio. Diferenças nos valores da matriz intrínseca, na posição do ponto principal, na distância focal efetiva e nos coeficientes de distorção são esperadas, pois refletem características construtivas e ópticas específicas de cada equipamento. Assim, a comparação entre a webcam e a segunda câmera reforçou a ideia de que os parâmetros de calibração são particulares para cada sistema de aquisição.

No experimento D, os resultados da correção de distorção mostraram a aplicação direta dos parâmetros obtidos nas etapas anteriores. A redução das deformações geométricas nas imagens confirmou que a calibração não é apenas um procedimento matemático de estimação, mas também uma etapa fundamental para melhorar a qualidade visual e a fidelidade geométrica das imagens em aplicações de visão computacional. A comparação entre cv.undistort() e cv.remap() mostrou que ambos os métodos são eficazes, embora o segundo ofereça maior flexibilidade em cenários de processamento repetitivo.

De forma geral, a sequência dos experimentos mostrou uma progressão coerente entre teoria, calibração prática e aplicação dos parâmetros estimados. Os estudos realizados evidenciaram que a calibração de câmeras é essencial quando se deseja obter medidas mais confiáveis, interpretar corretamente a geometria da cena e reduzir erros provocados por distorções ópticas. Além disso, o laboratório demonstrou que a qualidade dos resultados depende tanto do modelo matemático quanto do cuidado experimental adotado durante a aquisição dos dados.

## Conclusões
Conclui-se que o laboratório atingiu seus objetivos ao permitir a compreensão dos conceitos fundamentais de calibração de câmeras e sua aplicação prática em diferentes etapas experimentais. A partir dos estudos teóricos e das execuções realizadas, foi possível entender o papel dos parâmetros intrínsecos e extrínsecos, bem como a importância dos coeficientes de distorção para a modelagem mais fiel do sistema de imageamento.

Os resultados obtidos mostraram que a calibração é um procedimento essencial para aplicações de visão computacional que exigem maior precisão geométrica. A comparação entre diferentes câmeras evidenciou que cada dispositivo apresenta características próprias, o que torna necessário realizar calibração específica para cada sistema utilizado. Também foi verificado que a qualidade da calibração depende fortemente da diversidade e da qualidade das imagens adquiridas, incluindo fatores como foco, iluminação, resolução e variedade de poses do padrão.

Além disso, a etapa de correção de distorção demonstrou de forma concreta a utilidade prática da calibração, pois permitiu melhorar a fidelidade visual das imagens e reduzir deformações provocadas pela lente. Isso confirma que a calibração não deve ser vista apenas como uma etapa teórica ou matemática, mas como um recurso importante para aumentar a confiabilidade dos resultados em tarefas de medição, reconstrução geométrica e interpretação da cena.

Como limitações, destacam-se possíveis imprecisões decorrentes da qualidade das capturas, da quantidade de imagens utilizadas, do alinhamento do padrão de calibração e das características ópticas dos dispositivos empregados. Como melhoria futura, recomenda-se ampliar o número de imagens coletadas, diversificar ainda mais as poses do tabuleiro, controlar melhor as condições de iluminação e comparar os resultados com métricas quantitativas de erro reprojetado. Dessa forma, seria possível obter calibrações ainda mais robustas e análises comparativas mais completas.
